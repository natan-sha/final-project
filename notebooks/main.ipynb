#Stage 3. Preparing the files for merging

##first - the file Krankenhausverzeichnis_2023

```python
hospitals["kreis_code"] = (
    hospitals["Land"].astype(int).astype(str).str.zfill(2)
    + hospitals["Kreis"].astype(int).astype(str).str.zfill(3)
)
```

```python
per_kreis = (
    hospitals
    .groupby("kreis_code")["Adresse_Name"]     
    .nunique()                                  
    .reset_index(name="n_hospitals")           
)

print(per_kreis.shape)
```

##second - the cleaned version of the file population_by_age_series.csv

```python
pop_2023 = population[population["date"] == "31.12.2023"].copy()
```

```python
pop_2023 = pop_2023.copy()
for col in ["total_all", "total_male", "total_female",
            "german_all", "german_male", "german_female",
            "foreign_all", "foreign_male", "foreign_female"]:
    pop_2023[col] = pd.to_numeric(pop_2023[col], errors="coerce")
```

```python
pop_2023 = pop_2023[~pop_2023["kreis_code"].astype(str).str[:2].isin(["02", "11"])].copy() #excluding Berlin and Hamburg
```

```python
pop_2023 = pop_2023.dropna(subset=["total_all"]).copy()
print(pop_2023.shape)
```

```python
print(pop_2023["kreis_code"].nunique(), "unique Kreise")
```

```python
print("\nAge group categories:")
for ag in pop_2023["age_group"].unique():
    print(f"  {ag}")
```

```python
age_65_plus = [
    "65 bis unter 70 Jahre",
    "70 bis unter 75 Jahre",
    "75 bis unter 80 Jahre",
    "80 bis unter 85 Jahre",
    "85 bis unter 90 Jahre",
    "90 Jahre und mehr",
]
```

```python
total_pop = (
    pop_2023[pop_2023["age_group"] == "Insgesamt"]
    [["kreis_code", "kreis_name", "total_all"]]
    .rename(columns={"total_all": "pop_total"})
    .copy()
)
print(total_pop.shape)  
```

```python
pop_2023["total_all"] = pd.to_numeric(pop_2023["total_all"], errors="coerce")
```

```python
total_pop.head()
```

```python
pop_65 = (
    pop_2023[pop_2023["age_group"].isin(age_65_plus)]
    .groupby("kreis_code")["total_all"]
    .sum()
    .reset_index(name="pop_65_plus")
    )
print(pop_65.shape)
```

```python
pop_65.head()
```

```python
kreis_pop = total_pop.merge(pop_65, on="kreis_code")
```

```python
kreis_pop["share_65_plus"] = kreis_pop["pop_65_plus"] / kreis_pop["pop_total"]
print(kreis_pop.head())
```

```python
print("\nDistribution of share_65_plus:")
print(kreis_pop["share_65_plus"].describe())
```

#Stage 4 - the merge of the two datasets for 2023

```python
merged = kreis_pop.merge(per_kreis, on="kreis_code", how="left")
```

```python
merged["n_hospitals"] = merged["n_hospitals"].fillna(0).astype(int)
```

```python
merged["hospitals_per_100k"] = merged["n_hospitals"] / merged["pop_total"] * 100_000 #hospitals per 100 thousands people
```

```python
print(merged.shape)
```

```python
print(merged.head())
```

```python
print("\n=== Distribution of hospitals_per_100k ===")
print(merged["hospitals_per_100k"].describe())
```

```python
merged[merged["n_hospitals"] == 0]
```

#Stage 4. Plotting data for 2023

```python
import plotly.express as px

# --- 1. Building the "problem score" --------------------------------
merged["aging_rank"]  = merged["share_65_plus"].rank(ascending=True,  pct=True)   # 1.0 = biggest share of old population
merged["access_rank"] = merged["hospitals_per_100k"].rank(ascending=False, pct=True)  # 1.0 = worst access
merged["problem_score"] = merged["aging_rank"] + merged["access_rank"]
merged.loc[merged["n_hospitals"] == 0, "problem_score"] = merged["problem_score"].max() + 1 #zero access in 2 unique cases

# --- 2. Picking the "most problematic" 5 + the next 5 ------------------------------
top10 = merged.nlargest(10, "problem_score")
top5_codes  = top10.head(5)["kreis_code"].tolist()
next5_codes = top10.tail(5)["kreis_code"].tolist()

# --- 3. Tagging each Kreis with a category ---------------------------
def categorize(code):
    if code in top5_codes:  return "Most problematic (top 5)"
    if code in next5_codes: return "Problematic (next 5)"
    return "Other Kreise"

merged["category"] = merged["kreis_code"].apply(categorize)

# --- 4. Plot ------------------------------------------------------
color_map = {
    "Most problematic (top 5)": "#C8385C",   # cherry
    "Problematic (next 5)":     "#E89AAE",   # paler cherry
    "Other Kreise":             "#9AA5D9",   # muted blue
}

```

```python
fig = px.scatter(
    merged,
    x="share_65_plus",
    y="hospitals_per_100k",
    color="category",
    color_discrete_map=color_map,
    hover_name="kreis_name",
    trendline="ols",
    trendline_scope="overall",          # one trendline for all data, not per category
    trendline_color_override="gray",
    title="Hospital access vs. share of population aged 65+ — Germany, 2023",
    labels={
        "share_65_plus":      "Share of population aged 65+",
        "hospitals_per_100k": "Unique Hospitals per 100,000 residents",
        "category":           "",       # hide the legend title
    },
    opacity=0.7,
)

fig.update_traces(textposition="top center", textfont_size=10)


fig.add_annotation(
    text=("Source: Destatis Krankenhausverzeichnis 2023 · "
          "Regionalstatistik table 12411-03-03-4 (Bevölkerung nach Altersgruppen).<br>"
          "Hamburg and Berlin excluded due to Kreis-granularity mismatch."),
    xref="paper", yref="paper",
    x=0, y=-0.28,
    showarrow=False,
    font=dict(size=10, color="gray"),
    xanchor="left",
)
fig.update_layout(margin=dict(b=80)) 

fig.show()
```

```python
top5 = (
    merged[merged["category"] == "Most problematic (top 5)"]
    .sort_values("problem_score", ascending=False)
    [["kreis_name", "share_65_plus", "hospitals_per_100k"]]
)
print("Aging Kreise with lowest hospital access in 2023:")
print(top5.to_string(index=False))
```
